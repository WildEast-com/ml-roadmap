```python
import numpy as np

# строки = квартиры
# столбцы = [площадь, комнаты, этаж]
X = np.array([
    [30, 1, 2],
    [45, 2, 5],
    [60, 3, 7],
    [25, 1, 1],
    [80, 4, 10]
])

X
```

```python
import numpy as np
np.__version__
```

```python
X.shape
```

```python
import numpy as np

# средняя площадь
np.mean(X[:, 0])
```

```python
# максимальный этаж
np.max(X[:, 2])
```

```python
# средние значения по каждому признаку (по столбцам)
np.mean(X, axis=0)
```

```python
# Z-score стандартизация вручную (как StandardScaler)
X_mean = np.mean(X, axis=0)
X_std = np.std(X, axis=0)

X_z = (X - X_mean) / X_std
X_z
```

```python
np.mean(X_z, axis=0), np.std(X_z, axis=0)
```

```python
np.max(X[:, 0])
```

```python
np.max(X_z[:, 0])
```
